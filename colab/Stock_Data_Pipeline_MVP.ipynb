## 🕘 Trading Hours Detection

````python
def is_within_trading_hours() -> bool:
    """
    Check if current time is within Indian stock market trading hours
    NSE/BSE Trading Hours: 9:15 AM - 3:30 PM IST (Monday to Friday)
    """
    try:
        # Get current time in IST
        ist = pytz.timezone('Asia/Kolkata')
        now_ist = datetime.now(ist)
        
        # Check if it's a weekday (Monday = 0, Sunday = 6)
        if now_ist.weekday() >= 5:  # Saturday or Sunday
            return False
        
        # Market hours in IST: 9:15 AM to 3:30 PM
        market_open = now_ist.replace(hour=9, minute=15, second=0, microsecond=0)
        market_close = now_ist.replace(hour=15, minute=30, second=0, microsecond=0)
        
        # Check if current time is within trading hours
        is_trading_time = market_open <= now_ist <= market_close
        
        print(f"Current IST time: {now_ist.strftime('%Y-%m-%d %H:%M:%S %Z')}")
        print(f"Market open: {market_open.strftime('%H:%M')} IST")
        print(f"Market close: {market_close.strftime('%H:%M')} IST")
        print(f"Is trading hours: {is_trading_time}")
        
        return is_trading_time
        
    except Exception as e:
        print(f"Error checking trading hours: {e}")
        return True

# Test the function
print(f"🕘 Current time is within trading hours: {is_within_trading_hours()}")
````

## 📈 Mock Data Generation (Demo Mode)

````python
def get_mvp_symbol_list() -> List[str]:
    """Get a reduced list of symbols for MVP to stay within API limits"""
    mvp_symbols = [
        'NSE:RELIANCE-EQ',
        'NSE:TCS-EQ', 
        'NSE:HDFCBANK-EQ',
        'NSE:INFY-EQ',
        'NSE:ICICIBANK-EQ',
        'NSE:HINDUNILVR-EQ',
        'NSE:KOTAKBANK-EQ',
        'NSE:SBIN-EQ',
        'NSE:BHARTIARTL-EQ',
        'NSE:ITC-EQ'
    ]
    
    print(f"Using MVP symbol list with {len(mvp_symbols)} symbols")
    return mvp_symbols

def generate_mock_ohlcv_data(symbols: List[str]) -> Dict:
    """
    Generate realistic mock OHLCV data for demo mode
    This ensures the demo works even when API is down
    """
    print(f"🎬 Generating mock data for {len(symbols)} symbols")
    
    all_data = {}
    current_time = datetime.utcnow()
    
    for symbol in symbols:
        # Generate realistic price data
        base_price = random.uniform(100, 3000)  # Random base price
        
        # Generate 8 candles (30 minutes of 15-min data) 
        candles = []
        for i in range(8):
            # Each candle: [timestamp, open, high, low, close, volume]
            timestamp = int((current_time - timedelta(minutes=15*i)).timestamp())
            
            open_price = base_price + random.uniform(-5, 5)
            close_price = open_price + random.uniform(-10, 10)
            high_price = max(open_price, close_price) + random.uniform(0, 5)
            low_price = min(open_price, close_price) - random.uniform(0, 5)
            volume = random.randint(10000, 100000)
            
            candles.append([timestamp, open_price, high_price, low_price, close_price, volume])
        
        all_data[symbol] = {
            'symbol': symbol,
            'resolution': '15',
            'candles': candles,
            'timestamp': current_time.isoformat(),
            'total_records': len(candles),
            'mock_data': True  # Flag to indicate this is demo data
        }
        
        print(f"✅ Generated mock data for {symbol}: {len(candles)} candles")
    
    return {
        'data': all_data,
        'metadata': {
            'total_symbols_requested': len(symbols),
            'successful_symbols': len(symbols),  # All succeed in demo mode
            'failed_symbols': [],
            'success_rate_percent': 100.0,
            'ingestion_timestamp': current_time.isoformat(),
            'resolution': '15',
            'mvp_mode': True,
            'lambda_execution': True,
            'demo_mode': True,
            'data_source': 'mock_generated'
        }
    }

# Generate sample data
symbols = get_mvp_symbol_list()
sample_data = generate_mock_ohlcv_data(symbols)
print(f"\n📊 Generated data for {len(sample_data['data'])} symbols")
print(f"Success rate: {sample_data['metadata']['success_rate_percent']}%")
````

## 📊 Data Analysis & Visualization

````python
def analyze_ohlcv_data(data: Dict):
    """Analyze OHLCV data and display insights"""
    print("📈 OHLCV Data Analysis")
    print("=" * 50)
    
    for symbol, symbol_data in data['data'].items():
        if 'candles' in symbol_data:
            candles = symbol_data['candles']
            latest_candle = candles[-1]  # Most recent candle
            
            open_price = round(latest_candle[1], 2)
            high_price = round(latest_candle[2], 2)
            low_price = round(latest_candle[3], 2)
            close_price = round(latest_candle[4], 2)
            volume = int(latest_candle[5])
            
            # Calculate change
            change = round(close_price - open_price, 2)
            change_percent = round((change / open_price * 100), 2) if open_price > 0 else 0
            
            # Color coding for terminal
            change_color = "🟢" if change > 0 else "🔴" if change < 0 else "⚪"
            
            print(f"\n{symbol}:")
            print(f"  Open: ₹{open_price:,.2f}")
            print(f"  High: ₹{high_price:,.2f}")
            print(f"  Low:  ₹{low_price:,.2f}")
            print(f"  Close: ₹{close_price:,.2f}")
            print(f"  Volume: {volume:,}")
            print(f"  Change: {change_color} ₹{change:+.2f} ({change_percent:+.2f}%)")

# Analyze the sample data
analyze_ohlcv_data(sample_data)
````

## 🔄 Data to DataFrame Conversion

````python
def convert_to_dataframe(data: Dict) -> pd.DataFrame:
    """Convert OHLCV data to pandas DataFrame for analysis"""
    records = []
    
    for symbol, symbol_data in data['data'].items():
        if 'candles' in symbol_data:
            for candle in symbol_data['candles']:
                records.append({
                    'symbol': symbol,
                    'timestamp': pd.to_datetime(candle[0], unit='s'),
                    'open': float(candle[1]),
                    'high': float(candle[2]),
                    'low': float(candle[3]),
                    'close': float(candle[4]),
                    'volume': int(candle[5])
                })
    
    df = pd.DataFrame(records)
    return df

# Convert to DataFrame
df = convert_to_dataframe(sample_data)
print(f"📊 DataFrame created with {len(df)} records")
print(f"\nColumns: {list(df.columns)}")
print(f"\nSample data:")
print(df.head())

# Basic statistics
print(f"\n📈 Summary Statistics:")
print(df.groupby('symbol')[['open', 'high', 'low', 'close', 'volume']].mean().round(2))
````

## 🏗️ AWS Lambda Function (Production Code)

````python
def lambda_handler_demo(event, context):
    """
    AWS Lambda handler for stock data ingestion (Demo Version)
    This is the actual production code that runs in AWS Lambda
    """
    print("🚀 Starting MVP stock data ingestion (Demo Mode)")
    
    try:
        # Simulate environment variables
        demo_mode = True  # Always True for Colab demo
        
        print("🎬 DEMO MODE ENABLED - Using mock data")
        
        # Get symbol list
        symbols = get_mvp_symbol_list()
        print(f"Processing {len(symbols)} symbols for MVP")
        
        # Generate mock data (in production, this would call Fyers API)
        ohlcv_data = generate_mock_ohlcv_data(symbols)
        
        if not ohlcv_data or not ohlcv_data.get('data'):
            raise Exception("No data generated")
        
        # In production, this would upload to S3
        print("✅ Mock S3 upload successful")
        
        # Calculate metrics
        successful_symbols = ohlcv_data['metadata']['successful_symbols']
        total_symbols = ohlcv_data['metadata']['total_symbols_requested']
        success_rate = (successful_symbols / total_symbols * 100) if total_symbols > 0 else 0
        
        print(f"✅ MVP ingestion completed: {successful_symbols} symbols processed")
        
        return {
            'statusCode': 200,
            'body': {
                'message': 'Ingestion completed successfully',
                'symbols_processed': successful_symbols,
                'total_symbols': total_symbols,
                'success_rate': success_rate,
                'timestamp': datetime.utcnow().isoformat(),
                'demo_mode': True
            }
        }
        
    except Exception as e:
        error_msg = f"MVP stock data ingestion failed: {str(e)}"
        print(f"❌ {error_msg}")
        
        return {
            'statusCode': 500,
            'body': {
                'message': 'Ingestion failed',
                'error': str(e),
                'timestamp': datetime.utcnow().isoformat()
            }
        }

# Test the Lambda function
print("🧪 Testing Lambda Function:")
result = lambda_handler_demo({}, {})
print(f"\n📋 Lambda Result:")
print(json.dumps(result, indent=2, default=str))
````
